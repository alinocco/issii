# EDA: Anxiety survey dataset

Ноутбук для разведочного анализа данных опроса о тревожности.

Исходные данные ожидаются в папке `../data/raw` в формате CSV.

```python
import pandas as pd
from pathlib import Path

# Папка с сырыми CSV из Kaggle
raw_dir = Path("../data/raw")
csv_files = list(raw_dir.glob("*.csv"))
print("Найдено CSV-файлов:", len(csv_files))
csv_files
```

```python
# Загрузка одного CSV-файла в DataFrame

if not csv_files:
    raise FileNotFoundError("В папке ../data/raw не найдено ни одного CSV-файла. Сначала разместите там датасет с Kaggle.")

# при наличии нескольких файлов берём первый по имени
csv_path = sorted(csv_files)[0]
print(f"Используем файл: {csv_path}")

df = pd.read_csv(csv_path)
print(df.shape)
df.head()
```

```python
# Общая информация о данных

display(df.head())
print("\nИнформация о столбцах:")
print(df.info())

print("\nЧисловые признаки (describe):")
df.describe().T
```

```python
# Анализ пропусков

missing = df.isna().sum().sort_values(ascending=False)
missing = missing[missing > 0]
missing
```

```python
# Распределение целевой переменной (если есть)

target_cols = [c for c in df.columns if c.lower() in ["anxiety_binary", "anxiety", "target"]]

if not target_cols:
    print("Не найдена явная целевая колонка (anxiety_binary / anxiety / target). Проверьте имена столбцов в датасете.")
else:
    target_col = target_cols[0]
    print(f"Используем целевую колонку: {target_col}")
    display(df[target_col].value_counts(normalize=False).rename("count"))
    display(df[target_col].value_counts(normalize=True).rename("share"))
```

```python
# Простейшие визуализации

import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

# Гистограммы для нескольких числовых признаков
num_cols = df.select_dtypes(include=["number"]).columns.tolist()

if num_cols:
    sample_num_cols = num_cols[:6]
    df[sample_num_cols].hist(bins=20, figsize=(12, 8))
    plt.tight_layout()
    plt.show()
else:
    print("В датасете нет числовых признаков для построения гистограмм.")

# Корреляционная матрица по числовым признакам (если их достаточно)
if len(num_cols) >= 2:
    plt.figure(figsize=(10, 8))
    corr = df[num_cols].corr(numeric_only=True)
    sns.heatmap(corr, cmap="coolwarm", center=0)
    plt.title("Корреляционная матрица числовых признаков")
    plt.show()
```
